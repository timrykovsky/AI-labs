```python
# Лабораторная работа №1
# Предобработка данных. Знакомство с pandas

# 1. Подключение библиотек

import pandas as pd
from sklearn.preprocessing import MinMaxScaler


# 2. Загрузка датасета

df = pd.read_csv("train.csv")

print("Первые 5 строк датасета:")
print(df.head())


# 3. Первичный анализ датасета

print("\nИнформация о датасете:")
df.info()

print("\nТипы данных:")
print(df.dtypes)

print("\nНазвания столбцов:")
print(df.columns.tolist())


# 4. Поиск пропущенных значений

print("\nКоличество пропущенных значений:")
print(df.isnull().sum())


# 5. Заполнение пропущенных значений

# Заполняем пропуски в Age медианой
age_median = df["Age"].median()
df["Age"] = df["Age"].fillna(age_median)

# Заполняем пропуски в Embarked модой
embarked_mode = df["Embarked"].mode()[0]
df["Embarked"] = df["Embarked"].fillna(embarked_mode)

# Заполняем пропуски в Fare средним значением
fare_mean = df["Fare"].mean()
df["Fare"] = df["Fare"].fillna(fare_mean)

# Удаляем столбец Cabin,
# так как в нем очень много пропущенных значений
df = df.drop("Cabin", axis="columns")


# Проверяем, остались ли пропуски
print("\nКоличество пропущенных значений после обработки:")
print(df.isnull().sum())


# 6. Удаление ненужных для обработки столбцов

# Удаляем столбцы, которые не будем использовать
df = df.drop(["PassengerId", "Name", "Ticket"], axis="columns")

print("\nДатасет после удаления ненужных столбцов:")
print(df.head())


# 7. Нормализация числовых данных

# Список числовых столбцов,
# которые будем нормализовать
numeric_columns = [
    "Pclass",
    "Age",
    "SibSp",
    "Parch",
    "Fare"
]

# Создаём объект MinMaxScaler
scaler = MinMaxScaler()

# Выполняем нормализацию
df[numeric_columns] = scaler.fit_transform(df[numeric_columns])

print("\nЧисловые данные после нормализации:")
print(df[numeric_columns].head())


# 8. Преобразование категориальных данных

# Кодируем категориальные признаки методом One-Hot Encoding
# drop_first=True удаляет первый столбец каждой категории
df = pd.get_dummies(
    df,
    columns=["Sex", "Embarked"],
    drop_first=True
)

print("\nДатасет после преобразования категориальных данных:")
print(df.head())


# 9. Итоговая информация о датасете

print("\nИтоговая информация о датасете:")
df.info()

print("\nПервые 5 строк итогового датасета:")
print(df.head())


# 10. Сохранение обработанного датасета

df.to_csv("processed_titanic.csv", index=False)

print("\nОбработанный датасет сохранён в файл processed_titanic.csv")
```
